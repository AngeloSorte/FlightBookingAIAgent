## Flight search

### Subtask:
Utilizza le API di prenotazione voli per cercare voli in base ai parametri estratti dalla richiesta dell'utente.


**Reasoning**:
Define the `search_flights` function to interact with a hypothetical flight booking API.



```python
import requests
import os

def search_flights(origin, destination, departure_date, return_date=None, passengers=1):
    """
    Searches for flights using a hypothetical flight booking API.

    Args:
      origin: The departure airport code or city name.
      destination: The arrival airport code or city name.
      departure_date: The departure date in YYYY-MM-DD format.
      return_date: The return date in YYYY-MM-DD format (optional).
      passengers: The number of passengers.

    Returns:
      A list of dictionaries containing flight information or an empty list if
      no flights are found or an error occurs.
    """
    # In a real-world scenario, you would replace this with the actual API details
    # from the API integration step.
    base_url = "https://api.hypotheticalflightbookingapi.com/v1/flights"
    api_key = os.environ.get('FLIGHT_API_KEY', 'YOUR_HYPOTHETICAL_API_KEY') # Use environment variable or a placeholder

    headers = {
        "X-API-Key": api_key,
        "Content-Type": "application/json"
    }

    params = {
        "origin": origin,
        "destination": destination,
        "departure_date": departure_date,
        "passengers": passengers
    }

    if return_date:
        params["return_date"] = return_date

    try:
        print(f"Searching for flights: {params}") # For demonstration
        response = requests.get(base_url, headers=headers, params=params)
        response.raise_for_status()  # Raise an exception for bad status codes (4xx or 5xx)

        data = response.json()

        # In a real scenario, you would parse the API response structure
        # to extract relevant flight information. This is a placeholder.
        if data and data.get("success"):
            # Assuming the API returns a list of flights under a 'flights' key
            flights = data.get("flights", [])
            print(f"Found {len(flights)} flights.")
            return flights
        else:
            print("API response indicates no success or no flights found.")
            return []

    except requests.exceptions.RequestException as e:
        print(f"Error during API request: {e}")
        return []
    except ValueError as e:
        print(f"Error parsing API response: {e}")
        return []
```

## Flight selection and booking

### Subtask:
Permetti all'utente di selezionare un volo dai risultati della ricerca e procedi con la prenotazione tramite API.


**Reasoning**:
Define the `book_flight` function to simulate the flight booking process with a hypothetical API, including data preparation, making a POST request, handling errors, and parsing the response.



```python
import requests
import os

def book_flight(selected_flight, user_details):
  """
  Simulates booking a selected flight using a hypothetical booking API.

  Args:
    selected_flight: A dictionary containing the details of the selected flight.
                     Expected keys: 'flight_id', 'price', 'departure_time',
                     'arrival_time', 'origin', 'destination'.
    user_details: A dictionary containing user information for booking.
                  Expected keys: 'name', 'email', 'payment_info' (can be a dict).

  Returns:
    A dictionary with booking status and details, or an error message.
  """
  base_url = "https://api.hypotheticalflightbookingapi.com/v1/bookings"
  api_key = os.environ.get('FLIGHT_API_KEY', 'YOUR_HYPOTHETICAL_API_KEY') # Use environment variable or a placeholder

  headers = {
      "X-API-Key": api_key,
      "Content-Type": "application/json"
  }

  # Prepare data for the booking API
  booking_payload = {
      "flight_id": selected_flight.get("flight_id"),
      "passenger_details": {
          "name": user_details.get("name"),
          "email": user_details.get("email")
      },
      "payment_info": user_details.get("payment_info")
      # In a real scenario, payment_info would be more detailed
  }

  # Basic validation
  if not booking_payload["flight_id"] or not booking_payload["passenger_details"]["name"] or not booking_payload["passenger_details"]["email"]:
      return {"status": "failure", "message": "Missing required booking information."}


  try:
      print(f"Attempting to book flight with payload: {booking_payload}") # For demonstration
      response = requests.post(base_url, headers=headers, json=booking_payload)
      response.raise_for_status()  # Raise an exception for bad status codes (4xx or 5xx)

      booking_response_data = response.json()

      # Simulate parsing the API response
      # Assuming the API returns a JSON with a 'status' and 'booking_id' on success
      if booking_response_data.get("status") == "success":
          booking_id = booking_response_data.get("booking_id")
          print(f"Booking successful! Booking ID: {booking_id}")
          return {"status": "success", "booking_id": booking_id, "details": booking_response_data}
      else:
          # Assuming the API returns an error message in case of failure
          error_message = booking_response_data.get("message", "Unknown booking error.")
          print(f"Booking failed: {error_message}")
          return {"status": "failure", "message": error_message, "details": booking_response_data}

  except requests.exceptions.RequestException as e:
      print(f"Error during API request: {e}")
      return {"status": "failure", "message": f"API request failed: {e}"}
  except ValueError as e:
      print(f"Error parsing API response: {e}")
      return {"status": "failure", "message": f"Invalid API response: {e}"}
```

## Email generation and sending

### Subtask:
Una volta confermata la prenotazione, genera un'email con i dettagli del biglietto e inviala all'indirizzo specificato utilizzando un servizio di email.


**Reasoning**:
Define the `send_ticket_email` function to construct and send the email using `email.mime.text` and `smtplib`.



```python
import smtplib
from email.mime.text import MIMEText
import os

def send_ticket_email(booking_details, recipient_email):
    """
    Generates and sends an email with flight ticket details.

    Args:
      booking_details: A dictionary containing confirmed booking details.
                       Expected keys: 'booking_id', 'flight_id', 'departure_time',
                       'arrival_time', 'origin', 'destination', 'passenger_name'.
      recipient_email: The email address of the recipient.

    Returns:
      True if the email was sent successfully, False otherwise.
    """
    sender_email = os.environ.get('EMAIL_ADDRESS')
    sender_password = os.environ.get('EMAIL_PASSWORD')
    smtp_server = os.environ.get('SMTP_SERVER')
    smtp_port = int(os.environ.get('SMTP_PORT', 587)) # Default to 587 for TLS

    if not all([sender_email, sender_password, smtp_server, smtp_port]):
        print("Email configuration missing. Cannot send email.")
        return False

    # Construct the email content
    email_body = f"""
    Dear {booking_details.get('passenger_name', 'Passenger')},

    Your flight booking is confirmed!

    Booking ID: {booking_details.get('booking_id', 'N/A')}
    Flight ID: {booking_details.get('flight_id', 'N/A')}
    Origin: {booking_details.get('origin', 'N/A')}
    Destination: {booking_details.get('destination', 'N/A')}
    Departure Time: {booking_details.get('departure_time', 'N/A')}
    Arrival Time: {booking_details.get('arrival_time', 'N/A')}

    Thank you for booking with us.

    Best regards,
    Your AI Booking Agent
    """

    # Create the email message
    msg = MIMEText(email_body)
    msg['Subject'] = f"Your Flight Booking Confirmation (Booking ID: {booking_details.get('booking_id', 'N/A')})"
    msg['From'] = sender_email
    msg['To'] = recipient_email

    try:
        # Connect to the SMTP server
        # Use SSL for port 465, STARTTLS for other ports like 587
        if smtp_port == 465:
            server = smtplib.SMTP_SSL(smtp_server, smtp_port)
        else:
            server = smtplib.SMTP(smtp_server, smtp_port)
            server.starttls()  # Upgrade the connection to a secure encrypted SSL/TLS state

        # Log in to the email account
        server.login(sender_email, sender_password)

        # Send the email
        server.sendmail(sender_email, recipient_email, msg.as_string())

        # Close the connection
        server.quit()

        print(f"Email sent successfully to {recipient_email}")
        return True

    except Exception as e:
        print(f"Error sending email: {e}")
        return False
```
